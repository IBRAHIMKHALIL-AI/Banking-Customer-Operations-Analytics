```python
import pandas as pd
import re

# Load your cleaned customers file
df = pd.read_csv('customers.csv')

def normalize_phone(phone):
    # Handle nulls and float artifacts
    if pd.isna(phone):
        return "Missing"

    phone = str(phone).strip()

    # Catch those weird negative number artifacts (like -10680)
    if phone.startswith('-'):
        return "Invalid Data"

    ext = ""
    # Isolate extensions
    if 'x' in phone.lower():
        parts = phone.lower().split('x')
        phone = parts[0]
        ext = parts[1]

    # Strip everything except digits
    phone = re.sub(r'\D', '', phone)

    # Remove leading country codes (001 or 1)
    if phone.startswith('001') and len(phone) > 10:
        phone = phone[3:]
    elif phone.startswith('1') and len(phone) > 10:
        phone = phone[1:]

    # Reformat to standard structure
    if len(phone) == 10:
        formatted = f"({phone[:3]}) {phone[3:6]}-{phone[6:]}"
    elif len(phone) == 0:
        formatted = "Missing"
    else:
        formatted = phone # Fallback for bizarre lengths

    if ext:
        formatted += f" ext {ext}"

    return formatted

# Apply the function and save the clean file
df['Phone'] = df['Phone'].apply(normalize_phone)
df.to_csv('customers_cleaned.csv', index=False)
```
